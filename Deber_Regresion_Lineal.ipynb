```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('https://raw.githubusercontent.com/MiltonDavila2/CSV/refs/heads/main/tv-sales.csv')
```

```python
print(dataset)
```

```python
X = dataset.iloc[:,0].values
```

```python
print(X)
```

```python
Y = dataset.iloc[:,-1].values
```

```python
print(Y)
```

```python

from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state = 0)
```

```python

from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train.reshape(-1,1), Y_train)
```

```python
Y_pred_train = regressor.predict(X_train.reshape(-1,1))
Y_pred_test = regressor.predict(X_test.reshape(-1,1))
```

```python
#Visualizamos los resultados del training
plt.scatter(X_train,Y_train, color = 'red')
plt.plot(X_train, Y_pred_train, color = 'blue')
plt.title("TV  VS SALES")
plt.xlabel("TV (Publicidad cientos miles de dolares)")
plt.ylabel("Sales (Miles de unidades vendidas)")
plt.show()
```

```python
#Visualizamos los resultados del training
plt.scatter(X_test,Y_test, color = 'red')
plt.plot(X_test, Y_pred_test, color = 'blue')
plt.title("TV  VS SALES")
plt.xlabel("TV (Publicidad cientos miles de dolares)")
plt.ylabel("Sales (Miles de unidades vendidas)")
plt.show()
```

```python
r2_train = regressor.score(X_train.reshape(-1,1), Y_train)
r2_test = regressor.score(X_test.reshape(-1,1), Y_test)

print("R^2 (Training):", r2_train)
print("R^2 (Testing):", r2_test)
```
